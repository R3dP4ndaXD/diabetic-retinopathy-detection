```python
import os
import cv2
import numpy as np
from tqdm.notebook import tqdm
import time  # <-- THIS IS THE FIX

print("Libraries imported.")
```

```python
import os

# --- Configuration ---
# This is the correct path we found
INPUT_ROOT = '/kaggle/input/eyepacs-aptos-messidor-diabetic-retinopathy/augmented_resized_V2'

# These paths will now work
INPUT_TRAIN_DIR = os.path.join(INPUT_ROOT, 'train')
INPUT_VAL_DIR = os.path.join(INPUT_ROOT, 'val')

OUTPUT_ROOT = '/kaggle/working/processed_data'
OUTPUT_TRAIN_224 = os.path.join(OUTPUT_ROOT, 'train_224')
OUTPUT_VAL_224 = os.path.join(OUTPUT_ROOT, 'val_224')
OUTPUT_TRAIN_380 = os.path.join(OUTPUT_ROOT, 'train_380')
OUTPUT_VAL_380 = os.path.join(OUTPUT_ROOT, 'val_380')

print(f"Checking for train directory at: {INPUT_TRAIN_DIR}")
# This line will now succeed!
CLASSES = sorted([d for d in os.listdir(INPUT_TRAIN_DIR) if os.path.isdir(os.path.join(INPUT_TRAIN_DIR, d))])

# --- Create all output directories ---
print("Creating output directory structure...")
for output_dir in [OUTPUT_TRAIN_224, OUTPUT_VAL_224, OUTPUT_TRAIN_380, OUTPUT_VAL_380]:
    os.makedirs(output_dir, exist_ok=True)
    for class_name in CLASSES:
        os.makedirs(os.path.join(output_dir, class_name), exist_ok=True)

print(f"Output directories created in {OUTPUT_ROOT}")
print(f"Found classes: {CLASSES}")
```

```python
def process_images(input_dir, output_dir, size):
    """
    Loads all images from input_dir, resizes them, and saves to output_dir.
    Maintains the class subfolder structure.
    """
    print(f"\n--- Starting processing for {output_dir} (size: {size}x{size}) ---")
    start_time = time.time()
    total_images = 0
    
    # Iterate over each class folder ('0', '1', '2', '3', '4')
    for class_name in CLASSES:
        input_class_dir = os.path.join(input_dir, class_name)
        output_class_dir = os.path.join(output_dir, class_name)
        
        image_files = os.listdir(input_class_dir)
        
        # Use tqdm for a progress bar for each class
        for img_name in tqdm(image_files, desc=f"Class {class_name}"):
            try:
                img_path = os.path.join(input_class_dir, img_name)
                
                # Load image with cv2
                img = cv2.imread(img_path)
                if img is None:
                    print(f"Warning: Could not read {img_path}. Skipping.")
                    continue
                
                # Resize image
                # INTER_AREA is robust for shrinking images
                img_resized = cv2.resize(img, (size, size), interpolation=cv2.INTER_AREA)
                
                # Save the resized image to the new location
                save_path = os.path.join(output_class_dir, img_name)
                cv2.imwrite(save_path, img_resized)
                
                total_images += 1
                
            except Exception as e:
                print(f"Error processing {img_path}: {e}")
                
    end_time = time.time()
    print(f"--- Finished processing {total_images} images in {end_time - start_time:.2f} seconds ---")

print("Processing function defined.")
```

```python
# --- Run the preprocessing for all four combinations ---

# 1. Process 224x224 training images
process_images(INPUT_TRAIN_DIR, OUTPUT_TRAIN_224, 224)

# 2. Process 224x224 validation images
process_images(INPUT_VAL_DIR, OUTPUT_VAL_224, 224)

# 3. Process 380x380 training images
process_images(INPUT_TRAIN_DIR, OUTPUT_TRAIN_380, 380)

# 4. Process 380x380 validation images
process_images(INPUT_VAL_DIR, OUTPUT_VAL_380, 380)

print("\n\n✅ --- ALL IMAGE PREPROCESSING IS COMPLETE --- ✅")
```
